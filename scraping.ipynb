```python
# webスクレイピングに最低限必要なライブラリをインポート
import requests
from bs4 import BeautifulSoup
```

# HTTPリクエスト
## 基本のリクエスト

```python
# アクセスしたいWebサイトのURLを指定
url = 'https://www.musashino-u.ac.jp/'

# WebサーバーにHTTPリクエストを送信
# レスポンスを変数に格納しておく
res = requests.get(url)
```

```python
print(f"レスポンス：{res}")  # レスポンスオブジェクトを表示
print(f"レスポンスの型：{type(res)}")  # レスポンスオブジェクトの型を表示
print(f"ステータスコード：{res.status_code}")  # ステータスコードを表示
print(f"ステータスメッセージ：{res.reason}")  # ステータスメッセージを表示

print(f"リクエスト:{res.request}")  # リクエストオブジェクトを表示
print(f"リクエストの型：{type(res.request)}")  # リクエストオブジェクトの型を表示
print(f"リクエストヘッダー：{res.request.headers}")  # リクエストヘッダーを表示
print(f"リクエストメソッド：{res.request.method}")  # リクエストメソッドを表示
print(f"リクエストURL：{res.request.url}")  # リクエストURLを表示
```

```python
# レスポンス
print(f"レスポンスヘッダー：{res.headers}")  # レスポンスヘッダーを表示
print(f"レスポンスボディ：{res.text}")  # レスポンスボディを表示
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'
}
```

```python
# 別のURLにアクセスしてみる
# url = 'https://ja.wikipedia.org/wiki/%E6%AD%A6%E8%94%B5%E9%87%8E%E5%A4%A7%E5%AD%A6'
url = 'https://ja.wikipedia.org/wiki/武蔵野大学'
res = requests.get(url, headers=headers)
print(f"ステータスコード：{res.status_code}")  # ステータスコードを表示
print(f"レスポンスボディ：{res.text}")  # レスポンスボディを表示
```

## パラメータ付きのリクエスト

```python
#url = "https://www.musashino-u.ac.jp/search.html?query=データサイエンス"
url = "https://www.musashino-u.ac.jp/search.html"
params = {
    'query': 'データサイエンス'
}
res = requests.get(url, headers=headers, params=params)
res.encoding = res.apparent_encoding  # レスポンスのエンコーディングを設定
print(f"ステータスコード：{res.status_code}")  # ステータスコードを表示
print(f"レスポンスボディ：{res.text}")  # レスポンスボディを表示
```

## BeautifulSoupでHTMLソースから任意の情報を取得する

```python
url = "https://www.musashino-u.ac.jp/"
res = requests.get(url, headers=headers)
res.encoding = res.apparent_encoding  # レスポンスのエンコーディング
print(f"ステータスコード：{res.status_code}")  # ステータスコードを表示
```

```python
# HTMLのソースコードを解析
soup = BeautifulSoup(res.text, 'html.parser')
print(f"soupの型：{type(soup)}")  # soupオブジェクトの型を表示
```

```python
# ソースコード全体から<a>タグを抽出
soup_a = soup.a
print(f"<a>タグ：{soup_a}")  # 抽出した<a>タグを表示
print(f"<a>タグの型：{type(soup_a)}")  # 抽出した<a>タグの型を表示

#ソースコード全体から<p>タグを抽出
soup_p = soup.p
print(f"<p>タグ：{soup_p}")  # 最初に見つかった<p>タグを表示
print(f"<p>タグの型：{type(soup_p)}")  # <p>タグオブジェクトの型を表示
```

```python
#　ソースコード全体から<div>タグを抽出
soup_div = soup.div
#print(f"<div>タグ：{soup_div}")  # 抽出した<div>タグを表示
#print(f"<div>タグの型：{type(soup_div)}")  # 抽出した<div>タグの型を表示

soup_divs_div = soup_div.div
print(f"<div>タグの中の<div>タグ：{soup_divs_div}")  # 抽出した<div>タグの中の<div>タグを表示
print(f"<div>タグの中の<div>タグの型：{type(soup_divs_div)}")  # 抽出した<div>タグの中の<div>タグの型
```

```python
soup_div_div = soup.div.div
print(f"<div>タグの中の<div>タグ：{soup_div_div}")  # 抽出した<div>タグの中の<div>タグを表示
print(f"<div>タグの中の<div>タグの型：{type(soup_div_div)}")  # 抽出した<div>タグの中の<div>タグの型
```

```python
# Tagオブジェクト.stringでタグ内のテキストを取得できる
print(f"<a>タグのテキスト：{soup_a.string}")  # <a>タグ内のテキストを表示

print(f"<div>タグ内の<dig>タグのテキスト：{soup_div_div.string}") # <dig>タグ内のテキストを表示
```

## 指定したHTMLタグで要素を検索

```python
# 指定したタグが使われている要素のうち、最初に見つかった要素を取得する
h2 = soup.find('h2')
print(f"{h2}")  # 抽出した<h2>タグを表示
print(f"変数の型:{type(h2)}")  # 抽出した<h2>タグの型を表示
```

```python
h2_all = soup.find_all('h2')
print(f"{h2_all}")  # 抽出したすべての<h2>タグを表示
print(f"変数の型:{type(h2_all)}")  # 抽出したすべての<h2>タグの型を表示

for h2 in h2_all:
    if h2.span is not None:
        print(f"<h2>タグのテキスト:{h2.span.string}")  # 各<h2>タグ内の<span>タグのテキストを表示
```

```python
a_list = soup.find_all('a')
for a in a_list:
    print(f"{a.string}")

# 検索結果の数に制限をかける
a_list_limited = soup.find_all('a', limit=5)
for a in a_list_limited:
    print(f"{a.string}")
```

```python
# 複数のタグを指定して検索することも可能
# 例えば、<h2>タグと<h3>タグの両方を指定して検索する場合は以下のようにする
h2_h3_list = soup.find_all(['h2', 'h3'])
for tag in h2_h3_list:
    print(f"{tag.string}")  
```

## 任意の属性で検索

```python
tag_href_list = soup.find_all(attrs={"href":'/'},limit = 5)
for tag in tag_href_list:
    print(f"{tag}") # href属性が'/'のタグを表示
```

```python
# id属性やclass属性など、任意の属性で検索することも可能
# 例えば、id属性が'main-conrent'の要素を検索する場合は以下のようにする
tag_id_list = soup.find_all(id='top')
for tag in tag_id_list:
    print(f"{tag}") # id属性が'top'のタグを表示


tag_class_list = soup.find_all(class_ = 'p-notice',limit=5)
for tag in tag_class_list:
    print(f"{tag}") # class属性が'p-notice'のタグを表示
```

## CSSセレクタを使った検索

```python
# id属性を指定　( # + 検索ID名)
id_tags = soup.select('#top')
for tag in id_tags:
    print(f"{tag}") # id属性が'top'のタグを表示

## class属性を指定　( . + 検索クラス名)
class_tags = soup.select('.p-notice')
for tag in class_tags:
    print(f"{tag}") # class属性が'p-notice'のタグを表示

##HTML階層構造を指定
# 例えば、<div>タグの中の<h2>タグを検索する場合は以下のようにする
div_h2_tags = soup.select('div h2')
for tag in div_h2_tags:
    print(f"{tag}") # <div>タグの中の<h2>タグを表示


## ブラウザから取得した背テクタを使う
# 例えば、ブラウザの検証ツールで取得したセレクタが以下の場合:
# body > div.l-wrapper > main > div > div > section.l-top.l-top--features > article.p-numbers > div.p-numbers__column > div:nth-child(2) > div > div:nth-child(5) > div > div.c-numbers__no > span.js-countup
selector_tags = soup.select('body > div.l-wrapper > main > div > div > section.l-top.l-top--features > article.p-numbers > div.p-numbers__column > div:nth-child(2) > div > div:nth-child(5) > div > div.c-numbers__no > span.js-countup')
for tag in selector_tags:
    print(f"{tag}")  # 指定したセレクタに一致するタグを表示
```

## 外部サイトをスクレイピングする際のコーディングの注意点

```python
import time
```

```python
#for文等で特定のWebサイトに連続でアクセスする場合は、サーバーに負荷をかけないように適切な感覚をあけること
for i in range(5):
    time.sleep(1) #プログラムを1秒停止
    print(f"アクセス回数:{i+1}")
```

# 練習問題
下記のページから出演映画のタイトルを取得する
https://abehiroshi.la.coocan.jp/movie/eiga.htm

```python
# アクセスしたいWebサイトのURLを指定
kadai_url = 'https://abehiroshi.la.coocan.jp/movie/eiga.htm'

# HTMLを取得
response = requests.get(kadai_url)
response.encoding = response.apparent_encoding  # 文字化け対策

# BeautifulSoupで解析
soup = BeautifulSoup(response.text, "html.parser")

# strongタグをすべて抽出
titlles = soup.find_all("a",target="_blank")

# strongタグの中身だけ表示
for tag in titlles:
    time.sleep(1)  # サーバーに負荷をかけないように1秒待機
    print(f"{tag}")  
```
